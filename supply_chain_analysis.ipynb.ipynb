```python
df.rename(columns={
    'order date (dateorders)': 'order_date',
    'shipping date (dateorders)': 'shipping_date'
}, inplace=True)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipping_date'] = pd.to_datetime(df['shipping_date'])

df['delivery_days'] = (
    df['shipping_date'] - df['order_date']
).dt.days
```
